## Q6. Median value of horsepower


>##### 1. Find the median value of the horsepower column in the dataset.
>##### 2. Next, calculate the most frequent value of the same horsepower column.
>##### 3. Use the fillna method to fill the missing values in the horsepower column with the most frequent value from the previous step.
>##### 4. Now, calculate the median value of horsepower once again.

#### Has it changed?

```python
# Median value for original dataset
ori_median = df['horsepower'].median()

ori_median
```

```python
# Find the most frequent value of horsepower
most_hp_frequency = df['horsepower'].mode()[0]

most_hp_frequency
```

```python
# Verify with another method
df['horsepower'].value_counts().head()
```

```python
# Determine how many missing values
df['horsepower'].isnull().sum()
```

```python
# Fill the missing values with the most frequent value
df['horsepower'] = df['horsepower'].fillna(most_hp_frequency)
```

```python
# See if data updated 
df['horsepower'].value_counts().head()
```

```python
# Calculate the new median value
new_median = df['horsepower'].median()

new_median
```

```python
if new_median > ori_median:
    result = "increased"
elif new_median < ori_median:
    result = "decreased"
else:
    result = "no changed"

print(f"The new median {result}.")
```

## Q7. Sum of weights

>##### 1. Select all the cars from Asia
>##### 2. Select only columns vehicle_weight and model_year
>##### 3. Select the first 7 values
>##### 4. Get the underlying NumPy array. Let's call it X.
>##### 5. Compute matrix-matrix multiplication between the transpose of X and X. To get the transpose, use X.T. Let's call the result XTX.
>##### 6. Invert XTX.
>##### 7. Create an array y with values [1100, 1300, 800, 900, 1000, 1100, 1200].
>##### 8. Multiply the inverse of XTX with the transpose of X, and then multiply the result by y. Call the result w.


#### What's the sum of all the elements of the result?

```python
import numpy as np
```

```python
# Slice the requested data from the original dataset
new_df = df.loc[df['origin']=='Asia', ['vehicle_weight', 'model_year']].head(7)

new_df
```

```python
# Convert to numpy array
X = new_df.values

# OR use
# X = new_df.to_numpy()

X
```

```python
# Compute matrix-matrix multiplication 
XTX = X.T.dot(X)

XTX
```

```python
# Find the inverse
XTX_inverse = np.linalg.inv(XTX)

XTX_inverse
```

```python
# Define y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

y
```

```python
# Calculate the weight, w
w = XTX_inverse.dot(X.T).dot(y)

w
```

```python
# Compute the sum
w.sum().round(3)
```
